```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
data = {
    'hours_studied' : [1, 2, 3, 4, 5],
    'marks_scored' : [35, 45, 50, 65, 70]
}

df = pd.DataFrame(data)
X = df[['hours_studied']].values
y = df['marks_scored'].values

model = LinearRegression()
model.fit(X, y)

slope = model.coef_[0]
intercept = model.intercept_

#hours_to_predict = np.array([[6]])
predicted_marks = model.predict([[6]])

print(f"Slope Coefficient: {slope:.2f}")
print(f"Intercept: {intercept:.2f}")
print(f"Predicted Marks for 6 hours of study: {predicted_marks[0]:.2f}")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    'area': [1000, 1500, 1800, 2200, 2500],
    'bedrooms': [2, 3, 3, 4, 4],
    'price': [50, 65, 72, 85, 95]
}

df = pd.DataFrame(data)

#Independent features
X = df[['area', 'bedrooms']]
y = df['price']

#Training
lin_reg = LinearRegression()
lin_reg.fit(X,y)

coef_area = lin_reg.coef_[0]
coef_bedrooms = lin_reg.coef_[1]
intercept = lin_reg.intercept_

X_new = pd.DataFrame([[2000, 3]], columns=['area', 'bedrooms'])
predicted_price = lin_reg.predict(X_new)[0]

print(f"Intercept: {intercept:.4f}")
print(f"Area Coefficient: {coef_area:.4f}")
print(f"Bedrooms Coefficient: {coef_bedrooms:.4f}\n")
print(f"Predicted Price (2000 sq ft, 3 Bedrooms): {predicted_price:.2f} Lakhs")
```

Problem 3: Polynomial Regression
The following data points are collected from an experiment where the output y depends on input x.
x	y
1	2
2	8
3	18
4	32
5	50

Task: Fit a polynomial regression model of degree 2 to the given data. Report the fitted model equation and predict the value of y when x = 6.


```python
import numpy as np
from sklearn.linear_model import LinearRegression      # Used for both simple and polynomial regression
from sklearn.preprocessing import PolynomialFeatures   # Transforms features into polynomial terms
X = np.array([1, 2, 3, 5, 5]).reshape(-1, 1)
y = np.array([2, 8, 18, 32, 50])
poly_reg = PolynomialFeatures(degree=4)
X_poly = poly_reg.fit_transform(X)
lin_reg = LinearRegression()
lin_reg.fit(X_poly,y)
x_new = poly_reg.transform(np.array([[6]]))
y_pred = lin_reg.predict(x_new)

print(f"Coefficients: {lin_reg.coef_}")
print(f"Intercept: {lin_reg.intercept_}")
print(f"Predicted y at x = 6: {y_pred[0]:.2f}")
```
