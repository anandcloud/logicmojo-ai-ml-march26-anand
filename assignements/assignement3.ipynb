Q11. End-to-End ML Preprocessing Pipeline (Titanic)

Context: Prepare dataset for ML model training

Tasks:

1. Load dataset
2. Handle missing values
3. Encode categorical variables (sex, embarked)
4. Normalize numerical features
5. Convert to NumPy array

Output:

Final feature matrix ready for model training

```python
import numpy as np
import pandas as pd
import seaborn as sns
#ML Pipelines
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer 
from sklearn.linear_model import LogisticRegression

df_raw = sns.load_dataset("titanic")
df = df_raw.copy()
#df.isnull().sum()
# Fix missing values
df['age'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
#df.isnull().sum()
#encoding
data = pd.get_dummies(df, columns = ['sex', 'embarked'], drop_first = True)
#data.head()
#Drop columns
columns_keep = ['survived', 'age', 'fare', 'sex_male', 'embarked_Q', 'embarked_S']
data = data[columns_keep]
#data.head()
#Split
X = data.drop('survived', axis=1)
Y = data['survived']
#Train/ Test Split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)

#Normalise (Scaling)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

LogisticRegression().fit(X_train_scaled, Y_train)
```

Q12. Feature Selection using Statistical Understanding

Context: Not all features improve model performance.

Dataset: Iris

Tasks:

1. Compute correlation matrix
2. Identify highly correlated features
3. Drop redundant features
4. Explain:
Impact on overfitting and model performance

```python
import numpy as np
import seaborn as sns
df_raw = sns.load_dataset("iris")
df = df_raw.copy()
corr_matrix = df.drop(columns=['species']).corr()
print("Correlation Matrix: ")
print(corr_matrix)
```

```python
# Correlation greater than 90%
to_drop = []
#print(corr_matrix.columns)
for i in range(len(corr_matrix.columns)):
    for j in range(i):
        if abs(corr_matrix.iloc[i, j]) > 0.90:
            colname = corr_matrix.columns[i]
            to_drop.append(colname)
print(set(to_drop))
df_final = df.drop(columns=to_drop)
print(df_final.columns.tolist())
```

Q13. Custom Feature Transformation using apply()

Context: Real ML requires custom logic transformation.

Dataset: Titanic

Tasks:

1. Create age_group:
Child / Adult / Senior
2. Use .apply()
3. Compute survival rate per group
4. Interpret:
Which segment has highest survival likelihood

```python
import seaborn as sns
from pandas import isna
df_raw = sns.load_dataset("titanic")
df = df_raw.copy()
def cateogrise_age(age):
    if pd.isna(age):
        return "Unknown"
    elif age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

df['age_group'] = df['age'].apply(cateogrise_age)
survival_rates = df.groupby('age_group')['survived'].mean().sort_values(ascending=False).round(2)

print("Survival Rate by Age Group:")
print(survival_rates)
```

Q14. Error Analysis Simulation (Model Evaluation Thinking)

Context: Understanding prediction errors is key in ML.

Tasks:

1. Generate 1000 random error values using NumPy
2. Compute:
Mean
Standard deviation
3. Identify outliers:
Using mean ± 2*std
4. Explain:
What high variance means in model performance

```python
import numpy as np
error = np.random.normal(0, 5, 1000)
print("Mean:", np.mean(error).round(2))
print("Standard deviation: ", np.std(error).round())
limit = 2 * np.std(error)
outliers = error[np.abs(error - (np.mean(error))) > limit]
print(f"Number of 'Extreme' Errors: {len(outliers)}")
```
